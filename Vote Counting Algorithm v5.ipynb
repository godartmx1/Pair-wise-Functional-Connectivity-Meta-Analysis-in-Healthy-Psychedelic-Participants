```python
"""
Weighted vote-counting heatmaps for Yeo-7 functional-connectivity studies.

Included studies
----------------
- Bedford 2023
- Dai 2023
- de Araujo 2011
- Gaddis 2022
- Grimm 2018 (vote only; seed-to-voxel / no denominator workbook)
- Kaelen 2016 (vote only; seed-to-voxel / no denominator workbook)
- Madsen 2021
- Mallaroni 2024 (already in Yeo-7; weighted with native Yeo-pair denominators = 1)
- Mason 2020
- Palhano-Fontes 2015 (vote only; no denominator workbook)
- Pasquini 2020
- Roseman 2014
- Smigielski 2019
- Timmermann 2023 (already in Yeo-7; weighted with native Yeo-pair denominators = 1)

Weighting rule
--------------
For weighted-eligible studies, each significant edge contributes:

    +1 / Possible_Edges(pair)   if Direction indicates increased connectivity
    -1 / Possible_Edges(pair)   if Direction indicates decreased connectivity

Special cases
-------------
- Seed-to-voxel studies without a possible-edges workbook are included in vote matrices only.
- Native Yeo-7 studies can be weighted directly at the Yeo-pair level. For Mallaroni 2024,
  each Yeo-pair has one possible connection in the already-collapsed Yeo-7 representation,
  so Possible_Edges(pair) = 1 for every unique Yeo pair.
"""

from pathlib import Path
import math
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# =========================
# User paths
# =========================
BASE_DIR = Path("/Users/maximegodart/Desktop/CBMA_coordinate_extract/CONNECTIVITY/NEW ANALYSIS DATA")
OUT_DIR = BASE_DIR / "study_weighted_heatmaps"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# =========================
# Plot style
# =========================
sns.set_theme(
    style="white",
    context="talk",
    font="DejaVu Sans",
    rc={
        "figure.dpi": 150,
        "savefig.dpi": 300,
        "axes.titlesize": 16,
        "axes.labelsize": 12,
        "xtick.labelsize": 10,
        "ytick.labelsize": 10,
    },
)

# =========================
# Yeo-7 order
# =========================
YEO7_ORDER = [
    "Visual",
    "Somatomotor",
    "DorsalAttention",
    "VentralAttention",
    "Limbic",
    "Frontoparietal",
    "Default",
]
YEO2IDX = {k: i for i, k in enumerate(YEO7_ORDER)}

NETWORK_ALIASES = {
    "visual": "Visual",
    "vis": "Visual",
    "somatomotor": "Somatomotor",
    "sensorimotor": "Somatomotor",
    "somatormotor": "Somatomotor",
    "smn": "Somatomotor",
    "dorsalattention": "DorsalAttention",
    "dorsal attention": "DorsalAttention",
    "dan": "DorsalAttention",
    "ventralattention": "VentralAttention",
    "ventral attention": "VentralAttention",
    "salience": "VentralAttention",
    "van": "VentralAttention",
    "limbic": "Limbic",
    "frontoparietal": "Frontoparietal",
    "fronto parietal": "Frontoparietal",
    "fpn": "Frontoparietal",
    "executive": "Frontoparietal",
    "default": "Default",
    "defaultmode": "Default",
    "default mode": "Default",
    "dmn": "Default",
}

# =========================
# Study config
# =========================
STUDY_CONFIGS = [
    {
        "study": "Bedford 2023",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Bedford_2023_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Bedford_2023_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Dai 2023",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Dai_2023_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Dai_2023_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "de Araujo 2011",
        "converted_file": BASE_DIR / "data_yeo_converted" / "DeAraujo_2011_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "DeAraujo_2011_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Gaddis 2022",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Gaddis_2022_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Gaddis_2022_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Grimm 2018",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Grimm_2018_Yeo7_converted.xlsx",
        "converted_sheet": 0,
        "weight_mode": "vote_only",
        "weighting_note": "Seed-to-voxel study with MNI coordinates; excluded from weighted analysis.",
    },
    {
        "study": "Kaelen 2016",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Kaelen_2016_Yeo7_converted.xlsx",
        "converted_sheet": 0,
        "weight_mode": "vote_only",
        "weighting_note": "Seed-to-voxel study with MNI coordinates; excluded from weighted analysis.",
    },
    {
        "study": "Madsen 2021",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Madsen_2021_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Madsen_2021_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Mallaroni 2024",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Mallaroni_2024_Yeo7.xlsx",
        "converted_sheet": 0,
        "weight_mode": "native_yeo",
        "weighting_note": "Already in Yeo-7; weighted directly at the Yeo-pair level with Possible_Edges = 1 per unique Yeo pair.",
    },
    {
        "study": "Mason 2020",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Mason_2020_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Mason_2020_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Palhano-Fontes 2015",
        "converted_file": BASE_DIR / "data_yeo_converted" / "PalhanoFontes_2015_Yeo7_converted.xlsx",
        "converted_sheet": 0,
        "weight_mode": "vote_only",
        "weighting_note": "No possible-edges workbook available; included in vote analysis only.",
    },
    {
        "study": "Pasquini 2020",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Pasquini_2020_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Pasquini_2020_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Roseman 2014",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Roseman_2014_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Roseman_2014_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Smigielski 2019",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Smigielski_2019_Yeo7_converted.xlsx",
        "possible_edges_file": BASE_DIR / "possible_edge_outputs" / "Smigielski_2019_possible_edges_master.xlsx",
        "converted_sheet": 0,
        "possible_edges_sheet": "Master_possible_edges",
        "weight_mode": "possible_edges_file",
    },
    {
        "study": "Timmermann 2023",
        "converted_file": BASE_DIR / "data_yeo_converted" / "Timmermann_2023_Yeo7.xlsx",
        "converted_sheet": 0,
        "weight_mode": "native_yeo",
        "weighting_note": "Already in Yeo-7; weighted directly at the Yeo-pair level with Possible_Edges = 1 per unique Yeo pair.",
    },
]

# =========================
# Helpers
# =========================
def norm_text(x):
    if pd.isna(x):
        return ""
    x = str(x).strip()
    x = x.replace("&", "and")
    x = re.sub(r"[_\-]+", " ", x)
    x = re.sub(r"\s+", " ", x)
    return x.strip()


def canonical_network(x):
    raw = norm_text(x)
    key = raw.lower().replace(" ", "")
    if key in NETWORK_ALIASES:
        return NETWORK_ALIASES[key]

    raw_no_extra = raw.replace(" ", "")
    for canon in YEO7_ORDER:
        if raw_no_extra.lower() == canon.lower().replace(" ", ""):
            return canon

    raise ValueError(f"Unrecognized Yeo-7 network label: {x!r}")


def pair_key(a, b):
    a = canonical_network(a)
    b = canonical_network(b)
    return tuple(sorted((a, b), key=lambda z: YEO2IDX[z]))


def direction_to_sign(x):
    s = norm_text(x).lower()
    if s in {"increase", "increased", "positive", "hyperconnectivity", "greater", "up"}:
        return 1.0
    if s in {"decrease", "decreased", "negative", "hypoconnectivity", "less", "down"}:
        return -1.0
    if "increase" in s or "greater" in s or "hyper" in s or s == "+":
        return 1.0
    if "decrease" in s or "less" in s or "hypo" in s or s == "-":
        return -1.0
    raise ValueError(f"Could not infer sign from Direction value: {x!r}")


def safe_sum(mats, fill_value=0.0):
    if not mats:
        return np.full((len(YEO7_ORDER), len(YEO7_ORDER)), fill_value, dtype=float)
    out = np.full_like(mats[0], fill_value, dtype=float)
    for m in mats:
        out = out + m
    return out


def empty_matrix(fill=0.0):
    return np.full((len(YEO7_ORDER), len(YEO7_ORDER)), fill, dtype=float)


def matrix_to_df(mat):
    return pd.DataFrame(mat, index=YEO7_ORDER, columns=YEO7_ORDER)


def slugify(x):
    return re.sub(r"[^A-Za-z0-9]+", "_", str(x)).strip("_")


def read_excel_clean(path, sheet_name=0):
    df = pd.read_excel(path, sheet_name=sheet_name)
    df.columns = [str(c).strip() for c in df.columns]
    return df


def get_yeo_edge_networks(row):
    """
    Return canonical Yeo-7 labels for an edge row.

    Prefer reparcellated columns (Seed_Yeo7 / Target_Yeo7) when present, because some
    converted files retain the original seed/target labels in Seed / Target (e.g., Kaelen PHC).
    Fall back to Seed / Target for studies whose converted files already overwrite those columns
    with Yeo labels.
    """
    seed_val = row["Seed_Yeo7"] if "Seed_Yeo7" in row.index and pd.notna(row["Seed_Yeo7"]) and str(row["Seed_Yeo7"]).strip() != "" else row["Seed"]
    target_val = row["Target_Yeo7"] if "Target_Yeo7" in row.index and pd.notna(row["Target_Yeo7"]) and str(row["Target_Yeo7"]).strip() != "" else row["Target"]
    return canonical_network(seed_val), canonical_network(target_val)


def all_yeo_pair_denominators():
    rows = []
    pemap = {}
    for i, a in enumerate(YEO7_ORDER):
        for j, b in enumerate(YEO7_ORDER[i:]):
            b = YEO7_ORDER[i + j]
            k = pair_key(a, b)
            pemap[k] = 1.0
            rows.append({"A": k[0], "B": k[1], "Possible_Edges": 1.0})
    return pemap, pd.DataFrame(rows)


def load_possible_edges_map(xlsx_path, sheet_name="Master_possible_edges"):
    df = read_excel_clean(xlsx_path, sheet_name=sheet_name)

    required = {"Yeo_Pair_A", "Yeo_Pair_B", "Possible_Edges"}
    missing = required - set(df.columns)
    if missing:
        raise ValueError(f"Possible-edges file is missing required columns: {sorted(missing)}")

    pemap = {}
    debug_rows = []
    skipped_zero_rows = []

    for i, r in df.iterrows():
        a = canonical_network(r["Yeo_Pair_A"])
        b = canonical_network(r["Yeo_Pair_B"])
        pe = float(r["Possible_Edges"])

        if math.isnan(pe):
            raise ValueError(f"Row {i}: Possible_Edges is NaN")
        if pe <= 0:
            skipped_zero_rows.append({"row": i, "A": a, "B": b, "Possible_Edges": pe})
            continue

        k = pair_key(a, b)
        if k in pemap and not np.isclose(pemap[k], pe):
            raise ValueError(f"Conflicting Possible_Edges for pair {k}: {pemap[k]} vs {pe}")
        pemap[k] = pe
        debug_rows.append({"A": a, "B": b, "Possible_Edges": pe})

    debug_df = pd.DataFrame(debug_rows).sort_values(["A", "B"]).reset_index(drop=True)

    if skipped_zero_rows:
        skipped_df = pd.DataFrame(skipped_zero_rows).sort_values(["A", "B", "row"]).reset_index(drop=True)
        print(f"Note: skipped {len(skipped_df)} possible-edge rows with 0 edges in {Path(xlsx_path).name}")
        print(skipped_df.to_string(index=False))

    return pemap, debug_df


def load_zero_pairs(xlsx_path, sheet_name="Master_possible_edges"):
    pe_full = read_excel_clean(xlsx_path, sheet_name=sheet_name)
    zero_pairs = set()
    if {"Yeo_Pair_A", "Yeo_Pair_B", "Possible_Edges"}.issubset(pe_full.columns):
        for _, rr in pe_full.iterrows():
            try:
                aa = canonical_network(rr["Yeo_Pair_A"])
                bb = canonical_network(rr["Yeo_Pair_B"])
                pe0 = float(rr["Possible_Edges"])
            except Exception:
                continue
            if not math.isnan(pe0) and pe0 <= 0:
                zero_pairs.add(pair_key(aa, bb))
    return zero_pairs


def load_edges_df(converted_file, converted_sheet=0):
    edges = read_excel_clean(converted_file, sheet_name=converted_sheet)
    required_edge_cols = {"Seed", "Target", "Direction"}
    missing = required_edge_cols - set(edges.columns)
    if missing:
        raise ValueError(f"Converted Yeo-7 file is missing required columns: {sorted(missing)}")
    return edges


def compute_vote_only_study(converted_file, converted_sheet=0):
    edges = load_edges_df(converted_file, converted_sheet=converted_sheet)

    vote = empty_matrix(0.0)
    weighted = empty_matrix(0.0)
    row_debug = []

    for i, r in edges.iterrows():
        a, b = get_yeo_edge_networks(r)
        sign = direction_to_sign(r["Direction"])
        ia, ib = YEO2IDX[a], YEO2IDX[b]

        vote[ia, ib] += sign
        if ia != ib:
            vote[ib, ia] += sign

        row_debug.append({
            "row": i,
            "Seed_raw": r.get("Seed", np.nan),
            "Target_raw": r.get("Target", np.nan),
            "Seed_Yeo7_used": a,
            "Target_Yeo7_used": b,
            "Direction": r["Direction"],
            "sign": sign,
            "Possible_Edges": np.nan,
            "weighted_contribution": np.nan,
            "weight_status": "not_applicable_vote_only",
        })

    pe_debug = pd.DataFrame(columns=["A", "B", "Possible_Edges"])
    return vote, weighted, pd.DataFrame(row_debug), pe_debug, len(edges), len(edges)


def compute_weighted_from_map(converted_file, pemap, pe_debug, converted_sheet=0, zero_pairs=None):
    edges = load_edges_df(converted_file, converted_sheet=converted_sheet)
    zero_pairs = zero_pairs or set()

    vote = empty_matrix(0.0)
    weighted = empty_matrix(0.0)
    row_debug = []
    skipped_zero_pair_rows = []

    for i, r in edges.iterrows():
        a, b = get_yeo_edge_networks(r)
        sign = direction_to_sign(r["Direction"])
        k = pair_key(a, b)
        ia, ib = YEO2IDX[a], YEO2IDX[b]

        vote[ia, ib] += sign
        if ia != ib:
            vote[ib, ia] += sign

        if k in pemap:
            pe = pemap[k]
            weighted_contrib = sign / pe
            weighted[ia, ib] += weighted_contrib
            if ia != ib:
                weighted[ib, ia] += weighted_contrib
            weight_status = "used"
        elif k in zero_pairs:
            pe = 0.0
            weighted_contrib = np.nan
            weight_status = "skipped_zero_possible_edges"
            skipped_zero_pair_rows.append({
                "row": i,
                "Seed": a,
                "Target": b,
                "Direction": r["Direction"],
                "Yeo_pair": str(k),
            })
        else:
            raise ValueError(
                f"Row {i}: No Possible_Edges entry found for Yeo pair {k}.\n"
                f"Seed={r.get('Seed', np.nan)!r}, Target={r.get('Target', np.nan)!r}"
            )

        row_debug.append({
            "row": i,
            "Seed_raw": r.get("Seed", np.nan),
            "Target_raw": r.get("Target", np.nan),
            "Seed_Yeo7_used": a,
            "Target_Yeo7_used": b,
            "Direction": r["Direction"],
            "sign": sign,
            "Possible_Edges": pe,
            "weighted_contribution": weighted_contrib,
            "weight_status": weight_status,
        })

    if skipped_zero_pair_rows:
        skipped_df = pd.DataFrame(skipped_zero_pair_rows)
        print(
            f"Note: {len(skipped_df)} significant edge row(s) had Yeo pairs with 0 possible edges in "
            f"{Path(converted_file).name}; counted in VOTE heatmap but skipped in WEIGHTED heatmap."
        )
        print(skipped_df.to_string(index=False))

    return vote, weighted, pd.DataFrame(row_debug), pe_debug, len(edges), len(edges)


def compute_weighted_study_from_file(converted_file, possible_edges_file, converted_sheet=0, possible_edges_sheet="Master_possible_edges"):
    pemap, pe_debug = load_possible_edges_map(possible_edges_file, possible_edges_sheet)
    zero_pairs = load_zero_pairs(possible_edges_file, possible_edges_sheet)
    return compute_weighted_from_map(
        converted_file=converted_file,
        pemap=pemap,
        pe_debug=pe_debug,
        converted_sheet=converted_sheet,
        zero_pairs=zero_pairs,
    )


def compute_weighted_study_native_yeo(converted_file, converted_sheet=0):
    pemap, pe_debug = all_yeo_pair_denominators()
    return compute_weighted_from_map(
        converted_file=converted_file,
        pemap=pemap,
        pe_debug=pe_debug,
        converted_sheet=converted_sheet,
        zero_pairs=set(),
    )


def save_matrix_bundle(study_slug, vote, weighted, edge_debug, pe_debug):
    xlsx_path = OUT_DIR / f"{study_slug}_vote_weighted_outputs.xlsx"
    with pd.ExcelWriter(xlsx_path, engine="openpyxl") as writer:
        matrix_to_df(vote).to_excel(writer, sheet_name="vote_matrix")
        matrix_to_df(weighted).to_excel(writer, sheet_name="weighted_matrix")
        edge_debug.to_excel(writer, sheet_name="edge_debug", index=False)
        pe_debug.to_excel(writer, sheet_name="possible_edges_debug", index=False)
    return xlsx_path


def max_abs_for_scale(mats):
    vals = []
    for m in mats:
        arr = np.asarray(m, dtype=float)
        arr = arr[np.isfinite(arr)]
        if arr.size:
            vals.append(np.nanmax(np.abs(arr)))
    return max(vals) if vals else 1.0


def format_annotation(mat, fmt):
    ann = np.empty(mat.shape, dtype=object)
    for i in range(mat.shape[0]):
        for j in range(mat.shape[1]):
            v = mat[i, j]
            ann[i, j] = "NA" if not np.isfinite(v) else format(v, fmt)
    return ann


def plot_heatmap(mat, title, fmt=".0f", vmax=None, outstem=None):
    df = matrix_to_df(mat)
    ann = format_annotation(mat, fmt)

    if vmax is None or vmax <= 0 or not np.isfinite(vmax):
        finite_vals = np.asarray(mat, dtype=float)
        finite_vals = finite_vals[np.isfinite(finite_vals)]
        vmax = np.nanmax(np.abs(finite_vals)) if finite_vals.size else 1.0
        if vmax == 0:
            vmax = 1.0

    fig, ax = plt.subplots(figsize=(8.2, 7.0))
    hm = sns.heatmap(
        df,
        cmap="RdBu_r",
        vmin=-vmax,
        vmax=vmax,
        center=0,
        square=True,
        linewidths=0.6,
        linecolor="white",
        annot=ann,
        fmt="",
        annot_kws={"fontsize": 9},
        cbar_kws={"shrink": 0.85, "pad": 0.02},
        ax=ax,
    )

    ax.set_title(title, pad=12, fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.tick_params(axis="x", rotation=45)
    ax.tick_params(axis="y", rotation=0)

    cbar = hm.collections[0].colorbar
    cbar.set_label("Value", rotation=270, labelpad=18)

    # Bold diagonal labels slightly via grid emphasis
    for spine in ax.spines.values():
        spine.set_visible(False)

    fig.tight_layout()

    if outstem is not None:
        png_path = Path(str(outstem) + ".png")
        pdf_path = Path(str(outstem) + ".pdf")
        fig.savefig(png_path, dpi=300, bbox_inches="tight", facecolor="white")
        fig.savefig(pdf_path, dpi=300, bbox_inches="tight", facecolor="white")
    plt.show()


# =========================
# Main
# =========================
def main():
    results = []

    for cfg in STUDY_CONFIGS:
        study = cfg["study"]
        weight_mode = cfg.get("weight_mode", "possible_edges_file")

        print("\n" + "=" * 60)
        print(f"Processing: {study}")

        if weight_mode == "possible_edges_file":
            vote, weighted, edge_debug, pe_debug, n_used, n_total = compute_weighted_study_from_file(
                converted_file=cfg["converted_file"],
                possible_edges_file=cfg["possible_edges_file"],
                converted_sheet=cfg.get("converted_sheet", 0),
                possible_edges_sheet=cfg.get("possible_edges_sheet", "Master_possible_edges"),
            )
            include_in_weighted_meta = True
            atlas_path = str(cfg["possible_edges_file"])
        elif weight_mode == "native_yeo":
            note = cfg.get("weighting_note", "Native Yeo study weighted with Yeo-pair denominators.")
            print(f"Note: {note}")
            vote, weighted, edge_debug, pe_debug, n_used, n_total = compute_weighted_study_native_yeo(
                converted_file=cfg["converted_file"],
                converted_sheet=cfg.get("converted_sheet", 0),
            )
            include_in_weighted_meta = True
            atlas_path = "N/A (native Yeo-7 weighting)"
        elif weight_mode == "vote_only":
            note = cfg.get("weighting_note", "Weighting not applied.")
            print(f"Note: {note}")
            vote, weighted, edge_debug, pe_debug, n_used, n_total = compute_vote_only_study(
                converted_file=cfg["converted_file"],
                converted_sheet=cfg.get("converted_sheet", 0),
            )
            include_in_weighted_meta = False
            atlas_path = "N/A (vote only)"
        else:
            raise ValueError(f"Unknown weight_mode for {study}: {weight_mode!r}")

        study_slug = slugify(study)
        debug_xlsx = save_matrix_bundle(study_slug, vote, weighted, edge_debug, pe_debug)

        results.append({
            "study": study,
            "atlas_path": atlas_path,
            "n_edges_used": n_used,
            "n_edges_total": n_total,
            "vote": vote,
            "weighted": weighted,
            "include_in_weighted_meta": include_in_weighted_meta,
            "debug_xlsx": str(debug_xlsx),
            "weight_mode": weight_mode,
            "weighting_note": cfg.get("weighting_note", "") if weight_mode != "possible_edges_file" else "",
        })

    vote_meta = safe_sum([r["vote"] for r in results], fill_value=0.0)
    weighted_meta = safe_sum(
        [r["weighted"] for r in results if r["include_in_weighted_meta"]],
        fill_value=0.0,
    )

    vote_vmax = max_abs_for_scale([r["vote"] for r in results] + [vote_meta])
    weighted_vmax = max_abs_for_scale(
        [r["weighted"] for r in results if r["include_in_weighted_meta"]] + [weighted_meta]
    )

    for r in results:
        print("\n==============================")
        print(r["study"])
        print("Atlas:", r["atlas_path"])
        print("Debug workbook:", r["debug_xlsx"])
        print(f"Edges used: {r['n_edges_used']} / {r['n_edges_total']}")
        if r.get("weighting_note"):
            print("Weighted analysis:", r["weighting_note"])

        study_slug = slugify(r["study"])
        plot_heatmap(
            r["vote"],
            f"{r['study']} – Vote Heatmap",
            fmt=".0f",
            vmax=vote_vmax,
            outstem=OUT_DIR / f"{study_slug}_vote_heatmap",
        )

        if r["include_in_weighted_meta"]:
            plot_heatmap(
                r["weighted"],
                f"{r['study']} – Weighted Heatmap",
                fmt=".6f",
                vmax=weighted_vmax,
                outstem=OUT_DIR / f"{study_slug}_weighted_heatmap",
            )
        else:
            print(f"Skipping weighted heatmap for {r['study']} (not applicable).")

    meta_xlsx = OUT_DIR / "META_vote_weighted_outputs.xlsx"
    with pd.ExcelWriter(meta_xlsx, engine="openpyxl") as writer:
        matrix_to_df(vote_meta).to_excel(writer, sheet_name="vote_meta")
        matrix_to_df(weighted_meta).to_excel(writer, sheet_name="weighted_meta")
        pd.DataFrame(results)[[
            "study", "atlas_path", "n_edges_used", "n_edges_total", "include_in_weighted_meta",
            "weight_mode", "weighting_note", "debug_xlsx"
        ]].to_excel(writer, sheet_name="study_manifest", index=False)

    print("\n==============================")
    print("META (subset)")
    plot_heatmap(
        vote_meta,
        "META – Vote Heatmap",
        fmt=".0f",
        vmax=vote_vmax,
        outstem=OUT_DIR / "META_vote_heatmap",
    )
    plot_heatmap(
        weighted_meta,
        "META – Weighted Heatmap",
        fmt=".6f",
        vmax=weighted_vmax,
        outstem=OUT_DIR / "META_weighted_heatmap",
    )

    print("\nSaved outputs to:", OUT_DIR)
    print("Saved meta workbook:", meta_xlsx)


if __name__ == "__main__":
    main()
```
